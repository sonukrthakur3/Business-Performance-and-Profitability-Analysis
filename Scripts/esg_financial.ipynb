```python
# 1. Import Required Libraries
import pandas as pd
import numpy as np
```

```python
# 2. Load the Dataset
# (Make sure 'company_esg_financial_dataset.csv' is uploaded in Colab)
file_path = 'company_esg_financial_dataset.csv'
df = pd.read_csv(file_path)

print(" Data Loaded Successfully!")
print("-" * 50)
print(" Dataset Shape (Rows, Columns):", df.shape)
```

```python
# 3. Data Cleaning
print("\n🧹 Starting Data Cleaning...")

# Drop any duplicate rows if they exist
duplicates = df.duplicated().sum()
if duplicates > 0:
    df.drop_duplicates(inplace=True)
    print(f"🗑️ Dropped {duplicates} duplicate rows.")
```

```python
# Handle Missing Values (Filling numerical NaNs with 0 for safety in financial calc)
missing_values = df.isnull().sum().sum()
if missing_values > 0:
    df.fillna(0, inplace=True)
    print(f" Fixed {missing_values} missing values.")
else:
    print(" No missing values found!")
```

```python
# 4. Feature Engineering (Calculating Business KPIs)
print("\n Calculating Profit and Expenses...")

# Convert ProfitMargin to decimal if it is in percentage format (e.g., 15.5 instead of 0.155)
if df['ProfitMargin'].abs().mean() > 1:
    df['ProfitMargin'] = df['ProfitMargin'] / 100
```

```python
# Calculate Total Profit
df['Total_Profit'] = df['Revenue'] * df['ProfitMargin']

# Calculate Total Expenses
df['Total_Expenses'] = df['Revenue'] - df['Total_Profit']

print(" New Columns Added: 'Total_Profit' and 'Total_Expenses'")
```

```python
# 5. Quick Business Summary
print("-" * 50)
print(" QUICK BUSINESS SUMMARY:")
print(f"Total Revenue: ${df['Revenue'].sum():,.2f}")
print(f"Total Profit:  ${df['Total_Profit'].sum():,.2f}")
print(f"Total Expense: ${df['Total_Expenses'].sum():,.2f}")
print("-" * 50)
```

```python
# 6. Export Cleaned Data for Power BI
cleaned_file_name = "Cleaned_Business_Performance.csv"
df.to_csv(cleaned_file_name, index=False)

print(f" Success! Cleaned dataset saved as '{cleaned_file_name}'.")
print(" You can now download this file from Colab and load it into Power BI.")
```
